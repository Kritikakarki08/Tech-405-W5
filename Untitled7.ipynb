```python

import io
import json
import os
import random
import contextlib

import numpy as np
import torch
import torch.nn as nn
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix

# ----------------------------------------------------------------- config --
SEED = 42
HIDDEN = 128
LAYERS = 1
EPOCHS = 60
BATCH = 64
LR = 1e-3
BASELINE = 10.0   # % (from the hackathon sheet)
TARGET = 93.0     # % (from the hackathon sheet)
OUT = "outputs"
os.makedirs(OUT, exist_ok=True)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

log_lines = []


def log(*args):
    s = " ".join(str(a) for a in args)
    print(s)
    log_lines.append(s)


def text_snapshot(lines, path, title):
    """Render console-style text to a PNG so it can be used as a report snapshot."""
    n = len(lines)
    fig, ax = plt.subplots(figsize=(9, 0.32 * n + 0.9))
    ax.axis("off")
    fig.patch.set_facecolor("#1e1e1e")
    ax.text(0.0, 1.0, title, color="#9cdcfe", fontsize=11, family="monospace",
            va="top", fontweight="bold", transform=ax.transAxes)
    ax.text(0.0, 0.93, "\n".join(lines), color="#d4d4d4", fontsize=10,
            family="monospace", va="top", transform=ax.transAxes)
    fig.savefig(path, dpi=160, bbox_inches="tight", facecolor=fig.get_facecolor())
    plt.close(fig)


# ------------------------------------------------------------------- data --
d = load_digits()
X = d.images.astype(np.float32) / 16.0          # (1797, 8, 8), pixels 0..16 -> 0..1
y = d.target.astype(np.int64)

X_trainval, X_test, y_trainval, y_test = train_test_split(
    X, y, test_size=0.20, random_state=SEED, stratify=y)          # fixed 80/20
X_train, X_val, y_train, y_val = train_test_split(
    X_trainval, y_trainval, test_size=0.10, random_state=SEED, stratify=y_trainval)

majority = np.bincount(y_test).max() / len(y_test) * 100
data_lines = [
    f"images  d.images.shape        = {d.images.shape}   (samples, rows, cols)",
    f"each sample = sequence of 8 steps x 8 features  -> LSTM input (batch, 8, 8)",
    f"train   X {X_train.shape}  y {y_train.shape}",
    f"val     X {X_val.shape}   y {y_val.shape}   (10% of train, monitoring only)",
    f"test    X {X_test.shape}  y {y_test.shape}   (fixed 80/20 split, untouched)",
    f"classes = {sorted(set(y.tolist()))}",
    f"majority-class rate on test = {majority:.1f}%  (sheet baseline = {BASELINE:.0f}%)",
]
log("=== DATA ===")
for l in data_lines:
    log(l)

to_t = lambda a: torch.from_numpy(a)
Xtr, ytr = to_t(X_train).to(device), to_t(y_train).to(device)
Xva, yva = to_t(X_val).to(device), to_t(y_val).to(device)
Xte, yte = to_t(X_test).to(device), to_t(y_test).to(device)

# Snapshot 1: data + shapes (sample digits plus the shapes text)
fig = plt.figure(figsize=(10, 4.6))
gs = fig.add_gridspec(2, 10, height_ratios=[1, 1.1])
for i in range(10):
    ax = fig.add_subplot(gs[0, i])
    idx = np.where(y == i)[0][0]
    ax.imshow(d.images[idx], cmap="gray_r")
    ax.set_title(f"label {i}", fontsize=8)
    ax.axis("off")
axt = fig.add_subplot(gs[1, :])
axt.axis("off")
axt.text(0.0, 1.0, "\n".join(data_lines), family="monospace", fontsize=8.5, va="top")
fig.suptitle("Snapshot 1 - sklearn Digits: each 8x8 image is read as 8 time steps (rows) of 8 features",
             fontsize=10, fontweight="bold")
fig.tight_layout()
fig.savefig(f"{OUT}/snapshot1_data.png", dpi=160)
plt.close(fig)


# ------------------------------------------------------------------ model --
class DigitLSTM(nn.Module):
    def __init__(self, input_size=8, hidden=HIDDEN, layers=LAYERS, classes=10):
        super().__init__()
        self.lstm = nn.LSTM(input_size, hidden, num_layers=layers, batch_first=True)
        self.fc = nn.Linear(hidden, classes)

    def forward(self, x):                # x: (batch, 8 steps, 8 features)
        out, (h_n, c_n) = self.lstm(x)   # out: (batch, 8, hidden)
        return self.fc(h_n[-1])          # last hidden state -> logits (batch, 10)


model = DigitLSTM().to(device)
n_params = sum(p.numel() for p in model.parameters())
model_lines = str(model).splitlines() + [
    "",
    f"hidden={HIDDEN}  layers={LAYERS}  trainable parameters={n_params:,}",
    f"loss=CrossEntropyLoss  optimizer=Adam(lr={LR})  batch={BATCH}  epochs={EPOCHS}",
    "forward: LSTM -> last hidden state h_n[-1] -> Linear -> 10 logits",
]
log("\n=== MODEL ===")
for l in model_lines:
    log(l)
text_snapshot(model_lines, f"{OUT}/snapshot2_model.png",
              "Snapshot 2 - model definition (nn.LSTM -> nn.Linear)")

# --------------------------------------------------------------- training --
criterion = nn.CrossEntropyLoss()
opt = torch.optim.Adam(model.parameters(), lr=LR)


@torch.no_grad()
def evaluate(Xs, ys):
    model.eval()
    logits = model(Xs)
    loss = criterion(logits, ys).item()
    acc = (logits.argmax(1) == ys).float().mean().item() * 100
    return loss, acc, logits.argmax(1).cpu().numpy()


train_losses, val_losses, val_accs = [], [], []
best_val, best_state = -1.0, None
log("\n=== TRAINING ===")
for epoch in range(1, EPOCHS + 1):
    model.train()
    perm = torch.randperm(len(Xtr), device=device)
    total = 0.0
    for i in range(0, len(Xtr), BATCH):
        idx = perm[i:i + BATCH]
        opt.zero_grad()
        loss = criterion(model(Xtr[idx]), ytr[idx])
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step()
        total += loss.item() * len(idx)
    tr_loss = total / len(Xtr)
    va_loss, va_acc, _ = evaluate(Xva, yva)
    train_losses.append(tr_loss)
    val_losses.append(va_loss)
    val_accs.append(va_acc)
    if va_acc > best_val:
        best_val = va_acc
        best_state = {k: v.clone() for k, v in model.state_dict().items()}
    if epoch == 1 or epoch % 5 == 0:
        log(f"epoch {epoch:3d}/{EPOCHS}  train_loss {tr_loss:.4f}  "
            f"val_loss {va_loss:.4f}  val_acc {va_acc:.2f}%")

# Snapshot 3: training loss decreasing
fig, ax = plt.subplots(1, 2, figsize=(10, 3.8))
ax[0].plot(range(1, EPOCHS + 1), train_losses, label="train loss")
ax[0].plot(range(1, EPOCHS + 1), val_losses, label="val loss")
ax[0].set_xlabel("epoch"); ax[0].set_ylabel("cross-entropy loss"); ax[0].legend()
ax[0].set_title("Loss decreasing")
ax[1].plot(range(1, EPOCHS + 1), val_accs, color="tab:green")
ax[1].set_xlabel("epoch"); ax[1].set_ylabel("validation accuracy (%)")
ax[1].set_title("Validation accuracy")
fig.suptitle("Snapshot 3 - training: loss falls from "
             f"{train_losses[0]:.2f} to {train_losses[-1]:.3f} over {EPOCHS} epochs",
             fontsize=10, fontweight="bold")
fig.tight_layout()
fig.savefig(f"{OUT}/snapshot3_training.png", dpi=160)
plt.close(fig)

# ------------------------------------------------------------ final test --
model.load_state_dict(best_state)          # best epoch chosen on validation, not test
_, test_acc, preds = evaluate(Xte, yte)
cm = confusion_matrix(y_test, preds)
log("\n=== RESULT (test set, evaluated once) ===")
log(f"test accuracy = {test_acc:.2f}%   baseline = {BASELINE:.1f}%   target = {TARGET:.1f}%")
log("PASS: >= target (full marks)" if test_acc >= TARGET else
    "Below target" if test_acc > BASELINE else "At/below baseline")

fig, ax = plt.subplots(1, 2, figsize=(11, 4.4))
bars = ax[0].bar(["baseline", "target", "my LSTM"], [BASELINE, TARGET, test_acc],
                 color=["#999999", "#e69f00", "#0072b2"])
for b, v in zip(bars, [BASELINE, TARGET, test_acc]):
    ax[0].text(b.get_x() + b.get_width() / 2, v + 1, f"{v:.1f}%", ha="center")
ax[0].set_ylim(0, 105); ax[0].set_ylabel("test accuracy (%)")
ax[0].set_title("Test accuracy vs baseline & target")
im = ax[1].imshow(cm, cmap="Blues")
ax[1].set_xticks(range(10)); ax[1].set_yticks(range(10))
ax[1].set_xlabel("predicted"); ax[1].set_ylabel("true"); ax[1].set_title("Confusion matrix (test)")
for i in range(10):
    for j in range(10):
        if cm[i, j]:
            ax[1].text(j, i, cm[i, j], ha="center", va="center", fontsize=7,
                       color="white" if cm[i, j] > cm.max() / 2 else "black")
fig.suptitle(f"Snapshot 4 - final test accuracy = {test_acc:.2f}% on {len(y_test)} held-out images",
             fontsize=10, fontweight="bold")
fig.tight_layout()
fig.savefig(f"{OUT}/snapshot4_results.png", dpi=160)
plt.close(fig)

per_class = {str(c): float(cm[c, c] / cm[c].sum() * 100) for c in range(10)}
with open(f"{OUT}/metrics.json", "w") as f:
    json.dump({"dataset": "sklearn digits", "test_accuracy": test_acc,
               "baseline": BASELINE, "target": TARGET, "best_val_accuracy": best_val,
               "n_train": len(y_train), "n_val": len(y_val), "n_test": len(y_test),
               "params": n_params, "hidden": HIDDEN, "epochs": EPOCHS, "lr": LR,
               "batch": BATCH, "seed": SEED, "first_epoch_train_loss": train_losses[0],
               "final_train_loss": train_losses[-1], "per_class_accuracy": per_class},
              f, indent=2)
with open(f"{OUT}/train_log.txt", "w") as f:
    f.write("\n".join(log_lines))
```
